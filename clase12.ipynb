Escribir un programa para un dsPIC 4013 para que encienda un led a 3 segundos y otro a 11 segundos.

Conexiones recomendadas (Ambos LEDs conectados a Vss (GND) a través de una resistencia.)

LED1	RB0	330 Ω	LED encendido a los 3 s

LED2	RB1	330 Ω	LED encendido a los 11 s

```python
#include <xc.h>
#include <libpic30.h>   // Para usar __delay_ms() en dsPIC30F

// ================= CONFIGURACIÓN DE FUSES =================
_FOSC(CSW_FSCM_OFF & XT_PLL8);   // Oscilador XT con PLL x8
_FWDT(WDT_OFF);                  // Watchdog desactivado
_FBORPOR(PBOR_OFF & MCLR_EN);    // Habilita MCLR
_FGS(CODE_PROT_OFF);             // Sin protección de código

// ================= FRECUENCIA DEL RELOJ =================
#define FCY 8000000UL  // Frecuencia de instrucción (8 MHz si Fosc=32MHz con PLLx8)

// ================= PROGRAMA PRINCIPAL =================
int main(void) {

    // Configura los pines RB0 y RB1 como salidas
    TRISBbits.TRISB0 = 0;
    TRISBbits.TRISB1 = 0;

    // Inicializa LEDs en apagado
    LATBbits.LATB0 = 0;
    LATBbits.LATB1 = 0;

    // Espera 3 segundos (3000 ms)
    __delay_ms(3000);
    LATBbits.LATB0 = 1;   // Enciende LED1 (3 s)

    // Espera 8 segundos más (total 11 s desde inicio)
    __delay_ms(8000);
    LATBbits.LATB1 = 1;   // Enciende LED2 (11 s)

    // Queda en bucle infinito con los dos LEDs encendidos
    while(1);

    return 0;
}
```

A los 11 segundos: se enciende el LED2 (RB1).
